# Задача распознавания именованных сущностей (NER)

### **Цель занятия:**
Освоить основные подходы к решению задачи распознавания именованных сущностей (Named Entity Recognition, NER) в русскоязычных текстах.

### **Основная идея:**

Распознавание именованных сущностей (NER) - это задача обработки естественного языка, которая заключается в идентификации и классификации именованных сущностей в тексте по заранее определённым категориям: имена людей, названия организаций, географические названия, даты, денежные значения и т.д. NER является важным шагом во многих задачах NLP, таких как извлечение полезной информации, анализ тональности, машинный перевод и другие. Кроме того, на основании извлечённых сущностей можно строить различные графы зависимостей между ними.



### **Некоторые сложности в решении задачи NER**

Работа с русским языком накладывает некоторые ограничения на технологии распознавания именованных сущностей.

Особенности русского языка, усложняющие решение задачи NER:

* **Сложная морфология и словообразование**. В русском языке присутствуют различные падежи, окончания, склонения и спряжения. Одно имя или название может иметь десятки форм. Например, имя «Иван» может встречаться как «Ивана», «Ивану», «Иваном» и так далее. Получается, что идентифицировать сущность по простому совпадению со словом из словаря или шаблоном не всегда возможно.

* **Ограниченность размеченных данных в отдельных доменах**. Большинство качественных обучающих выборок для задач NLP — на английском языке. Для русского языка часто не хватает больших и разнообразных датасетов с точной разметкой сущностей. Эффективность обучения моделей на ограниченных наборах для некоторых областей применения в данном случае снижается по сравнению с результатами для англоязычных текстов.

Также сложности могут быть вызваны особенностями данных из конкретной области. Так, у текстов в социальных сетях есть следующие особенности , усложняющие решение задачи NER:

* **Капитализация (регистр) как отдельный признак**. Имена и названия часто пишут с маленькой буквы, что лишает нас возможности использовать один из самых очевидных признаков именованных сущностей.

* **Отсутствие пунктуации**. Пунктуация в таких текстах часто носит произвольный характер, что усложняет разделение сообщений на логические части и формирование связей между словами в пределах выделенного контекста.

* **Сленг**. Как и в других задачах NLP, сленг и редкая/нестандартная лексика снижают точность распознавания.



### **Практические случаи использования именованных сущностей**

Технология распознавания именованных сущностей находит применение во множестве практических сценариев. Вот несколько примеров:

* Автоматизация работы с документацией — системы электронного документооборота используют NER для извлечения ключевых данных из договоров, счетов, резюме и других документов.

* Мониторинг СМИ и социальных сетей — отслеживание упоминаний брендов, персон и событий для анализа репутации и рынка.

* Биомедицинские исследования — извлечение информации о взаимодействиях генов, белков и лекарственных препаратов из научных публикаций.

* Юридический анализ — автоматическая обработка судебных решений, законопроектов и правовых документов.

* Персональные цифровые ассистенты — распознавание команд пользователя, содержащих именованные сущности.

* Аналитика клиентских обращений — выявление часто упоминаемых продуктов, проблем и конкурентов в отзывах и запросах поддержки.

В каждой области применения методов NER есть свои особенности. Как минимум, будут различаться наборы классов, которые мы хотим идентифицировать. Поэтому невозможно предложить универсальный подход к решению данной задачи. Очень часто это будет комбинация из разных методов, например, использование правил и словарей для выделения основных сущностей с последующим применением модели BERT для обработки более сложных случаев.

### Основные типы именованных сущностей:

* PERSON — имена людей (Александр Пушкин, Мария Склодовская-Кюри).
* ORGANIZATION — названия компаний, учреждений, правительственных органов (Яндекс, ООН, МГУ).
* LOCATION — географические объекты (Москва, Волга, Альпы).
* DATE — даты и периоды времени (2 мая 2025 года, XVIII век).
* TIME — время суток (14:30, полдень).
* MONEY — денежные суммы (150 рублей, $1 миллион).
* PERCENT — проценты и доли (25%, одна треть).
* PRODUCT — названия продуктов и товаров (iPhone 15, Лада Веста).
* EVENT — названия событий (Олимпийские игры, Вторая мировая война).
* WORKOFART — названия произведений искусства (Война и мир, Мона Лиза).
* LAW — названия законов и нормативных актов (Конституция РФ).
* LANGUAGE — названия языков (русский, Python).
* QUANTITY — физические величины с единицами измерения (5 километров, 100 кг).

# Библиотека SpaCy

Перед началом работы необходимо установить библиотеку SpaCy и загрузить предобученную модель для русского языка.

Загрузим установленную русскую модель SpaCy и обработаем следующий простейший пример текста с часто встречающимися словами.

```python
import spacy

# Загрузка русской модели (самой маленькой из доступных)
nlp = spacy.load("ru_core_news_sm")

# Пример текста на русском языке
text = "Владимир Путин встретился с Сергеем Лавровым в Москве."

# Обработка текста
doc = nlp(text)

# Вывод распознанных сущностей
for ent in doc.ents:
    print(f"Сущность: {ent.text}, Тип: {ent.label_}")
```

Мы видим, что модель сразу объединяет токены, которые относятся к одной сущности (имени человека), в спаны - неразделимые куски текста.

В данном примере всего 2 типа: PERSON (PER) — имена людей и LOCATION — географические объекты. Все слова определены корректно.

Рассмотрим ещё один пример.

```python
text_2 = "Компания Google объявила о запуске нового продукта 15 октября 2023 года. Офис находится в Калифорнии."

# Обработка текста
doc_2 = nlp(text_2)

# Вывод распознанных сущностей
for ent in doc_2.ents:
    print(f"Сущность: {ent.text}, Тип: {ent.label_}")
```

*   **Морфология русского языка:** Различные склонения и падежи могут усложнять распознавание, хотя SpaCy и учитывает морфологию.

# Использование предобученных трансформеров в задаче NER

На этом занятии просто попробуем предобученную модель на наших примерах (https://huggingface.co/Gherman/bert-base-NER-Russian)

Она использует расширенную схему разметки **BILOU** -
Begin, Inside, Last, Other, Unit, рассмотренную на лекции.

B: Начало сущности (первое слово)

I: Продолжение сущности

O: Другие слова

L: Конец сущности (последнее слово)

U: однословная сущность



```python
from transformers import pipeline

ner_pipe = pipeline("ner", model="Gherman/bert-base-NER-Russian")
```

```python
text = "Меня зовут Сергей Иванович из Москвы."
results = ner_pipe(text)

for result in results:
    print(f"Word: {result['word']}, Entity: {result['entity']}, Score: {result['score']:.4f}")
```

```python
text = "18 февраля в Саудовской Аравии состоялись первые за три года полноценные переговоры делегаций России и США. Они были посвящены урегулированию на Украине, а также восстановлению отношений Москвы и Вашингтона. "
results = ner_pipe(text)

for result in results:
    print(f"Word: {result['word']}, Entity: {result['entity']}, Score: {result['score']:.4f}")
```

Мы видим, что модель верно предсказывает сущности, однако она разбивает незнакомые слова на отдельные знакомые токены. Это немного усложняет последующий анализ.

**Принцип работы**

Токенами служат слова, доступные в словаре, или их составные части. Если слово отсутствует в словаре, оно разбивается на части, которые в словаре присутствуют.

Например: Слово «играть» (playing) разбито на части: «play» и «##ing».
Префикс «##» показывает модели, что этот токен — не начало нового слова, а продолжение предыдущего.
